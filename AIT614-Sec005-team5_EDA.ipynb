```python
##I defined the NOIR datatypes that is what columns are what datatype, using python and pandas 
# loop over the columns in the DataFrame and determine their NOIR data type
for col in austin_df.columns:
    dtype = str(austin_df[col].dtype)
    if CategoricalDtype.is_dtype(austin_df[col]):
        dtype = 'Nominal'
    elif austin_df[col].dtype == int:
        dtype = 'Ordinal' if austin_df[col].nunique() < 10 else 'Integer'
    elif austin_df[col].dtype == float:
        dtype = 'Ratio'
    print(col + ': ' + dtype)
```

```python
import pandas as pd

# Assuming austin_df is your DataFrame

# Handle missing values
# For numerical columns, replace missing values with mean for continuous variables and median for ordinal variables
numerical_columns = ['id', 'host_id', 'neighbourhood', 'latitude', 'longitude', 
                     'price', 'minimum_nights', 'number_of_reviews', 
                     'reviews_per_month', 'calculated_host_listings_count', 
                     'availability_365', 'number_of_reviews_ltm']

for col in numerical_columns:
    if austin_df[col].dtype == 'int64' or (austin_df[col] % 1 == 0).all():
        austin_df[col].fillna(austin_df[col].mean(), inplace=True)  # For continuous variables, fill with mean
    else:
        austin_df[col].fillna(austin_df[col].median(), inplace=True)  # For ordinal variables, fill with median

# For categorical columns, replace missing values with mode (most frequent value)
categorical_columns = ['name', 'host_name', 'room_type', 'booking_date']

for col in categorical_columns:
    if austin_df[col].dtype == 'object':
        austin_df[col].fillna(austin_df[col].mode()[0], inplace=True)

# Data type conversion
# Convert columns with 'Ratio' data type to numerical
ratio_columns = ['id', 'host_id', 'neighbourhood', 'price', 'minimum_nights', 
                 'number_of_reviews', 'reviews_per_month', 'calculated_host_listings_count', 
                 'availability_365', 'number_of_reviews_ltm']

for col in ratio_columns:
    austin_df[col] = pd.to_numeric(austin_df[col], errors='coerce')

# Now, you have a cleaned austin_df ready for visualization and machine learning
```

```python
import folium
from folium.plugins import MarkerCluster
import numpy as np  # Ensure numpy is imported for handling NaN checks

# Create a map centered around Austin
austin_map = folium.Map(location=[30.2672, -97.7431], zoom_start=11)

# Add marker cluster to map
marker_cluster = MarkerCluster().add_to(austin_map)

# Plot each point using the austin_df DataFrame
for idx, row in austin_df.iterrows():
    if not np.isnan(row['latitude']) and not np.isnan(row['longitude']):
        folium.Marker(location=[row['latitude'], row['longitude']],
                      popup=f"{row['name']}: ${row['price']}").add_to(marker_cluster)

# Display the map
austin_map
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Convert 'booking_date' to datetime if not already
austin_df['booking_date'] = pd.to_datetime(austin_df['booking_date'], errors='coerce')  # Use 'coerce' to handle any non-datetime convertible errors

# Group by month-year and sum reviews per month
monthly_reviews = austin_df.groupby(austin_df['booking_date'].dt.to_period("M"))['reviews_per_month'].sum()

plt.figure(figsize=(14, 7))
monthly_reviews.plot(kind='line', marker='o', linestyle='-')
plt.title('Monthly Booking Trends Based on Reviews')  # Updated title
plt.xlabel('Month-Year')
plt.ylabel('Total Bookings per Month')
plt.ylim(1, 200)  # Set the limits of the Y-axis from 1 to 200
plt.show()  # Ensure to add plt.show() to display the plot
```

```python
import matplotlib.pyplot as plt

# Convert neighbourhood column to integer and remove decimal places
austin_df['neighbourhood'] = austin_df['neighbourhood'].astype(int)

plt.figure(figsize=(12, 8))
listing_counts = austin_df['neighbourhood'].value_counts()
bar_plot = listing_counts.plot(kind='bar', color='skyblue')  # Default color for all bars
plt.title('Listing Counts by Neighbourhood')
plt.xlabel('Neighbourhood')
plt.ylabel('Number of Listings')
plt.xticks(rotation=90)

# Assigning custom colors to bars based on neighbourhoods
colors = ['blue', 'green', 'red', 'orange', 'purple', 'yellow', 'brown', 'pink', 'grey', 'cyan']  # Define custom colors
for i, bar in enumerate(bar_plot.patches):
    bar.set_color(colors[i % len(colors)])  # Loop through bars and assign colors from the defined list
    # Adding text labels to each bar
    label = listing_counts.iloc[i]  # Accessing the count using .iloc for correct positioning
    plt.text(bar.get_x() + bar.get_width() / 2, bar.get_height(), str(label),
             ha='center', va='bottom', fontsize=9, rotation=90)

plt.show()
```

```python
plt.figure(figsize=(8, 6))
room_type_counts = austin_df['room_type'].value_counts()
room_type_counts.plot(kind='bar', color='Orange')

plt.title('Room Type Distribution')
plt.xlabel('Room Type')
plt.ylabel('Count')
plt.xticks(rotation=0)  # Rotate x-axis labels if necessary
plt.grid(axis='y')  # Add grid lines for better readability

# Add text labels for each bar
for i, count in enumerate(room_type_counts):
    plt.text(i, count + 10, str(count), ha='center', va='bottom')

plt.show()
```
